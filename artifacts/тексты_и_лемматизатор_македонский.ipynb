```python
nlp = classla.Pipeline('mk', processors='tokenize,pos,lemma')

test = "Мачката јаде голема риба во езерото."
doc = nlp(test)

for sentence in doc.sentences:
    for word in sentence.words:
        print(f"{word.text:15} → лемма: {word.lemma:15} | POS: {word.upos}")
```

```python
import os

preprocessed_folder = '/content/preprocessed_texts/'
os.makedirs(preprocessed_folder, exist_ok=True)

def prepare_macedonian_text(text):
    chunk_size = 5000
    chunks = [text[i:i+chunk_size] for i in range(0, len(text), chunk_size)]

    result_tokens = []
    for chunk in chunks:
        if not chunk.strip():
            continue
        doc = nlp(chunk)
        for sentence in doc.sentences:
            for word in sentence.words:
                if word.upos == 'PUNCT':
                    continue
                elif word.upos == 'PROPN':
                    result_tokens.append('person1')
                elif word.upos == 'PRON':
                    result_tokens.append('pron1')
                elif word.upos == 'NUM':
                    result_tokens.append('ordinal1')
                else:
                    result_tokens.append(word.lemma.lower())

    return ' '.join(result_tokens)

all_files = sorted([f for f in os.listdir(texts_folder) if f.endswith('.txt')])
print(f"Файлов для обработки: {len(all_files)}")

for i, fname in enumerate(all_files):
    out_path = os.path.join(preprocessed_folder, fname)

    if os.path.exists(out_path):
        print(f"[{i+1}/{len(all_files)}] Пропускаю (уже готов): {fname}")
        continue

    in_path = os.path.join(texts_folder, fname)
    with open(in_path, 'r', encoding='utf-8') as f:
        text = f.read()

    print(f"[{i+1}/{len(all_files)}] Обрабатываю: {fname} ({len(text):,} символов)...", end=' ')
    processed = prepare_macedonian_text(text)

    with open(out_path, 'w', encoding='utf-8') as f:
        f.write(processed)

    print(f"{len(processed.split()):,} токенов")
```
