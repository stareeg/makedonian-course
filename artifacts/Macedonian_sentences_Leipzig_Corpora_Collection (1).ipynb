1. Любую ссылку из вашего списка ниже можно открыть напрямую, например:
https://downloads.wortschatz-leipzig.de/corpora/mkd_news_2020_100K.tar.gz

После скачивания: mkd_news_2020_100K.tar.gz

Распакуй:

tar -xzf mkd_news_2020_100K.tar.gz

Внутри будет файл: mkd_news_2020_100K-sentences.txt

Это готовый корпус предложений.

Пример содержимого:

1   Владата денеска донесе нов закон .
2   Претседателот изјави дека економијата расте .

2. Скачать через Python и сохранить

Можно автоматически скачать и сохранить предложения.

import requests
import tarfile
import io

url = "https://downloads.wortschatz-leipzig.de/corpora/mkd_news_2020_100K.tar.gz"

r = requests.get(url)

tar = tarfile.open(fileobj=io.BytesIO(r.content), mode="r:gz")
tar.extractall("macedonian_corpus")

Теперь файл лежит на диске:

macedonian_corpus/
   mkd_news_2020_100K/
       mkd_news_2020_100K-sentences.txt

3. Сохранить только предложения (без ID)

Можно сразу сделать чистый текст:

sentences = []

with open(
    "macedonian_corpus/mkd_news_2020_100K/mkd_news_2020_100K-sentences.txt",
    encoding="utf8"
) as f:
    for line in f:
        parts = line.strip().split("\t")
        if len(parts) == 2:
            sentences.append(parts[1])

with open("macedonian_sentences.txt", "w", encoding="utf8") as f:
    for s in sentences:
        f.write(s + "\n")

Получится файл: macedonian_sentences.txt

С содержимым:

Владата денеска донесе нов закон .
Претседателот изјави дека економијата расте .
...
